```python
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
model_roberta = SentenceTransformer('all-roberta-large-v1')
```

```python
model_minilm_l12 = SentenceTransformer('sentence-transformers/all-MiniLM-L12-v2')
```

```python
input_file = 'biobert_embedding_terms.csv'  # Replace with the path to your input CSV file
output_file_roberta = 'roberta.csv'  # Replace with the desired output CSV file name
output_file_minilm_l12='all-MiniLM-L12-v2.csv'
```

```python
data = pd.read_csv(input_file)
```

```python
# Generate embeddings for each tumor name
embeddings_roberta = data['Tumor_Names'].apply(lambda x: model_roberta.encode(x))

# Convert embeddings into a DataFrame
embeddings_roberta_df = pd.DataFrame(embeddings_roberta.tolist(), columns=[f"Dim_{i}" for i in range(len(embeddings_roberta[0]))])
```

```python
embeddings_minilm_l12 = data['Tumor_Names'].apply(lambda x: model_minilm_l12.encode(x))

# Convert embeddings into a DataFrame
embeddings_minilm_l12_df = pd.DataFrame(embeddings_minilm_l12.tolist(), columns=[f"Dim_{i}" for i in range(len(embeddings_minilm_l12[0]))])

```

```python
output_data_roberta = pd.concat([data['Tumor_Names'], embeddings_roberta_df], axis=1)
output_data_minilm_l12 = pd.concat([data['Tumor_Names'], embeddings_minilm_l12_df], axis=1)
```

```python
output_data_roberta.to_csv(output_file_roberta, index=False)
```
