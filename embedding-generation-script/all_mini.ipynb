```python
# Load the pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

```

```python
model2=SentenceTransformer('all-mpnet-base-v2')
```

```python
model3 = SentenceTransformer('sentence-transformers/gtr-t5-large')
```

```python
model = SentenceTransformer('e5-large')
```

```python
# Load the input CSV file
input_file = 'biobert_embedding_terms.csv'  # Replace with the path to your input CSV file
output_file = 'all-MiniLM-L6-v2.csv'  # Replace with the desired output CSV file name
output_file2='all-mpnet-base-v2.csv'
output_file3='gtr-t5-large.csv'
```

```python
data = pd.read_csv(input_file)
```

```python
# Generate embeddings for each tumor name
embeddings = data['Tumor_Names'].apply(lambda x: model.encode(x))

# Convert embeddings into a DataFrame
embeddings_df = pd.DataFrame(embeddings.tolist(), columns=[f"Dim_{i}" for i in range(len(embeddings[0]))])

```

```python
embeddings2 = data['Tumor_Names'].apply(lambda x: model2.encode(x))

# Convert embeddings into a DataFrame
embeddings_df2 = pd.DataFrame(embeddings2.tolist(), columns=[f"Dim_{i}" for i in range(len(embeddings2[0]))])

```

```python
embeddings3 = data['Tumor_Names'].apply(lambda x: model3.encode(x))
embeddings_df3 = pd.DataFrame(embeddings3.tolist(), columns=[f"Dim_{i}" for i in range(len(embeddings3[0]))])
```

```python
# Concatenate the tumor names and embeddings
output_data = pd.concat([data['Tumor_Names'], embeddings_df], axis=1)

```

```python
output_data2 = pd.concat([data['Tumor_Names'], embeddings_df2], axis=1)
```

```python
output_data3 = pd.concat([data['Tumor_Names'], embeddings_df3], axis=1)
```
